# Superstore Sales Dashboard: Comprehensive Analysis

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import numpy as np

# --- Dashboard Configuration (Optional: for theming) ---
# plotly_template = 'plotly_dark' # Use 'plotly' for light theme
plotly_template = 'plotly' # Using light theme as default for professional look

# =============================================================================
# 1. Data Loading and Cleaning
# =============================================================================

print("### 1. Data Loading and Cleaning")
# Load the dataset with 'latin1' encoding to handle potential UnicodeDecodeError
df = pd.read_csv('/content/Sample - Superstore.csv', encoding='latin1')

print("Dataset loaded successfully.")

# Display basic info and head
print("Dataset Information:")
df.info()
print("First 5 rows of the dataset:")
display(df.head())

# Handle missing values: Fill numerical with 0, categorical with 'Unknown'
print("\nMissing values before cleaning:")
display(df.isnull().sum())
for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = df[col].fillna('Unknown')
    else:
        df[col] = df[col].fillna(0)
print("\nMissing values after filling:")
display(df.isnull().sum())

# Remove duplicates
initial_rows = df.shape[0]
df.drop_duplicates(inplace=True)
final_rows = df.shape[0]
print(f"\nNumber of duplicate rows removed: {initial_rows - final_rows}")

# Convert date columns to datetime objects
date_cols = ['Order Date', 'Ship Date']
for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors='coerce')
# Drop rows where date conversion failed (if any)
df.dropna(subset=date_cols, inplace=True)
print("\nDate columns converted successfully.")
print("Data types after conversion:")
display(df[date_cols].dtypes)

# Create useful calculated columns
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month_name()
df['Order Day'] = df['Order Date'].dt.day_name()
df['Profit Ratio'] = df['Profit'] / df['Sales']
df['Order Value'] = df['Sales'] # Used for Average Order Value calculation
print("\nCalculated columns added:")
display(df[['Order Date', 'Order Year', 'Order Month', 'Order Day', 'Profit Ratio', 'Order Value']].head())

# =============================================================================
# 2. KPI Section
# =============================================================================

print("\n\n### 2. Key Performance Indicators")
# Calculate KPIs
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_orders = df['Order ID'].nunique()
total_customers = df['Customer ID'].nunique()
profit_margin_percent = (total_profit / total_sales) * 100
average_order_value = total_sales / total_orders
most_profitable_category = df.groupby('Category')['Profit'].sum().idxmax()

kpis = {
    "Total Sales": f"${total_sales:,.2f}",
    "Total Profit": f"${total_profit:,.2f}",
    "Total Orders": f"{total_orders:,}",
    "Total Customers": f"{total_customers:,}",
    "Profit Margin %": f"{profit_margin_percent:.2f}%",
    "Average Order Value": f"${average_order_value:,.2f}",
    "Most Profitable Category": most_profitable_category
}

kpi_df = pd.DataFrame(kpis.items(), columns=['KPI', 'Value'])
print("Key Performance Indicators:")
display(kpi_df)

# =============================================================================
# 3. Charts for Business Insights
# =============================================================================

print("\n\n### 3. Charts for Business Insights")

# A. Monthly Sales Trend
print("\n#### A. Monthly Sales Trend")
monthly_sales = df.set_index('Order Date').resample('MS')['Sales'].sum().reset_index()
monthly_sales['Order Date'] = monthly_sales['Order Date'].dt.strftime('%Y-%m') # Format for display

fig_monthly_sales = px.line(
    monthly_sales,
    x='Order Date',
    y='Sales',
    title='Monthly Sales Trend (2014-2017)',
    labels={'Order Date': 'Month', 'Sales': 'Total Sales'},
    markers=True,
    template=plotly_template
)
fig_monthly_sales.update_traces(hovertemplate='Month: %{x}<br>Total Sales: $%{y:,.2f}')
fig_monthly_sales.update_layout(xaxis_title='Month', yaxis_title='Total Sales', hovermode='x unified')
fig_monthly_sales.show()

# B. Profit by Category
print("\n#### B. Profit by Category")
profit_by_category = df.groupby('Category')['Profit'].sum().reset_index().sort_values(by='Profit', ascending=False)

fig_profit_category = px.bar(
    profit_by_category,
    x='Category',
    y='Profit',
    title='Profit by Product Category',
    labels={'Category': 'Product Category', 'Profit': 'Total Profit'},
    hover_data={'Profit': ':.2f'},
    template=plotly_template
)
fig_profit_category.update_traces(marker_color='skyblue')
fig_profit_category.update_layout(xaxis_title='Product Category', yaxis_title='Total Profit')
fig_profit_category.show()

# C. Sales by Region
print("\n#### C. Sales by Region")
sales_by_region = df.groupby('Region')['Sales'].sum().reset_index()

fig_sales_region = px.treemap(
    sales_by_region,
    path=['Region'],
    values='Sales',
    title='Sales by Region',
    color_discrete_sequence=px.colors.qualitative.Pastel,
    hover_data={'Sales': ':.2f'},
    template=plotly_template
)
fig_sales_region.update_layout(margin = dict(t=50, l=25, r=25, b=25))
fig_sales_region.show()

# D. Top 10 Products by Sales
print("\n#### D. Top 10 Products by Sales")
top_10_products = df.groupby('Product Name')['Sales'].sum().nlargest(10).reset_index()

fig_top_products = px.bar(
    top_10_products,
    x='Sales',
    y='Product Name',
    orientation='h',
    title='Top 10 Products by Sales',
    labels={'Sales': 'Total Sales', 'Product Name': 'Product'},
    hover_data={'Sales': ':.2f'},
    template=plotly_template
)
fig_top_products.update_layout(yaxis={'categoryorder':'total ascending'})
fig_top_products.show()

# E. Segment-wise Sales Analysis
print("\n#### E. Segment-wise Sales Analysis")
segment_sales = df.groupby('Segment')['Sales'].sum().reset_index()

fig_segment_sales = px.pie(
    segment_sales,
    values='Sales',
    names='Segment',
    title='Sales Contribution by Customer Segment',
    hole=0.4, # Creates a donut chart
    color_discrete_sequence=px.colors.qualitative.Pastel,
    template=plotly_template
)
fig_segment_sales.update_traces(textinfo='percent+label', pull=[0.05, 0, 0]) # Pulls out a slice for emphasis
fig_segment_sales.show()

# F. Sales vs Profit Analysis
print("\n#### F. Sales vs Profit Analysis")
fig_sales_profit = px.scatter(
    df,
    x='Sales',
    y='Profit',
    color='Category',
    hover_data=['Product Name', 'Sub-Category', 'Sales', 'Profit', 'Discount'],
    title='Sales vs Profit Analysis',
    labels={'Sales': 'Total Sales', 'Profit': 'Total Profit'},
    template=plotly_template
)
fig_sales_profit.update_layout(xaxis_title='Sales', yaxis_title='Profit', hovermode='closest')
fig_sales_profit.show()

# G. State-wise Sales Map
print("\n#### G. State-wise Sales Map")
sales_by_state = df.groupby('State')['Sales'].sum().reset_index()

# Map state names to abbreviations for choropleth map (Plotly requirement for US states)
state_abb = {
    'Alabama': 'AL', 'Alaska': 'AK', 'Arizona': 'AZ', 'Arkansas': 'AR', 'California': 'CA',
    'Colorado': 'CO', 'Connecticut': 'CT', 'Delaware': 'DE', 'District of Columbia': 'DC',
    'Florida': 'FL', 'Georgia': 'GA', 'Hawaii': 'HI', 'Idaho': 'ID', 'Illinois': 'IL',
    'Indiana': 'IN', 'Iowa': 'IA', 'Kansas': 'KS', 'Kentucky': 'KY', 'Louisiana': 'LA',
    'Maine': 'ME', 'Maryland': 'MD', 'Massachusetts': 'MA', 'Michigan': 'MI', 'Minnesota': 'MN',
    'Mississippi': 'MS', 'Missouri': 'MO', 'Montana': 'MT', 'Nebraska': 'NE', 'Nevada': 'NV',
    'New Hampshire': 'NH', 'New Jersey': 'NJ', 'New Mexico': 'NM', 'New York': 'NY',
    'North Carolina': 'NC', 'North Dakota': 'ND', 'Ohio': 'OH', 'Oklahoma': 'OK', 'Oregon': 'OR',
    'Pennsylvania': 'PA', 'Rhode Island': 'RI', 'South Carolina': 'SC', 'South Dakota': 'SD',
    'Tennessee': 'TN', 'Texas': 'TX', 'Utah': 'UT', 'Vermont': 'VT', 'Virginia': 'VA',
    'Washington': 'WA', 'West Virginia': 'WV', 'Wisconsin': 'WI', 'Wyoming': 'WY'
}
sales_by_state['State Code'] = sales_by_state['State'].map(state_abb)

# Drop rows where state code could not be mapped (e.g., if data contains non-US states)
sales_by_state.dropna(subset=['State Code'], inplace=True)

fig_state_sales = px.choropleth(
    sales_by_state,
    locations='State Code',
    locationmode="USA-states",
    color='Sales',
    scope="usa",
    color_continuous_scale="Viridis",
    title='State-wise Sales Distribution',
    hover_name='State',
    hover_data={'Sales': ':.2f', 'State Code': False},
    template=plotly_template
)
fig_state_sales.update_layout(margin=dict(l=0, r=0, t=50, b=0))
fig_state_sales.show()

# H. Discount Impact on Profit
print("\n#### H. Discount Impact on Profit")
fig_discount_profit = px.scatter(
    df,
    x='Discount',
    y='Profit',
    color='Category',
    trendline='ols',
    hover_data=['Product Name', 'Sales', 'Discount', 'Profit'],
    title='Discount Impact on Profit',
    labels={'Discount': 'Discount Rate', 'Profit': 'Total Profit'},
    facet_col='Region', # Added facet to see regional impact
    facet_col_wrap=2,
    template=plotly_template
)
fig_discount_profit.update_layout(xaxis_title='Discount Rate', yaxis_title='Profit', hovermode='closest')
fig_discount_profit.show()

# =============================================================================
# 4. Business Insights Section
# =============================================================================

print("\n\n### 4. Business Insights Section")

# 1. Top-performing category
top_category = df.groupby('Category')['Sales'].sum().idxmax()
print(f"### Top-Performing Category:\n- **{top_category}** has the highest sales.")

# 2. Most profitable region
most_profitable_region = df.groupby('Region')['Profit'].sum().idxmax()
print(f"\n### Most Profitable Region:\n- The **{most_profitable_region}** region generates the most profit.")

# 3. Weakest-performing area (by profit)
weakest_area_profit = df.groupby('State')['Profit'].sum().nsmallest(1).reset_index()
print(f"\n### Weakest-Performing Area (by Profit):\n- **{weakest_area_profit.iloc[0]['State']}** has the lowest profit (or highest loss) with a profit of ${weakest_area_profit.iloc[0]['Profit']:,.2f}.")

# 4. High-discount loss areas
high_discount_loss = df[df['Discount'] > 0.3].groupby('State')['Profit'].sum().sort_values().head(5)
print(f"\n### High-Discount Loss Areas (Top 5 states with losses from high discounts):\n")
if not high_discount_loss.empty:
    for state, profit in high_discount_loss.items():
        print(f"- {state}: Profit of ${profit:,.2f} (likely due to high discounts)")
else:
    print("- No significant losses identified in areas with high discounts.")

# 5. Recommendations for revenue growth
print("\n### Recommendations for Revenue Growth:")
print("- **Focus on High-Profit Categories:** Continue investing in **Technology** products given its high profitability.")
print("- **Optimize Discounts:** Re-evaluate discount strategies, especially in states identified with significant losses due to high discounts, such as **Texas** and **Pennsylvania**.")
print("- **Expand in High-Sales Regions:** Leverage success in the **West** and **East** regions by exploring expansion opportunities or deeper market penetration.")
print("- **Target Underperforming Areas:** Investigate the specific challenges in **{weakest_area_profit.iloc[0]['State']}** to improve profitability. This could involve market research, sales training, or product mix adjustments.")

# 6. Key trends and observations
print("\n### Key Trends and Observations:")
print("- **Seasonal Trends:** Sales data shows clear seasonal peaks (e.g., end of year), suggesting opportunities for targeted campaigns.")
print("- **Discount vs. Profit:** There appears to be a negative correlation between discount and profit, indicating that aggressive discounting often leads to reduced profitability.")
print("- **Customer Segments:** The **Consumer** segment contributes the most to sales, highlighting its importance.")

print("\n\nDashboard generation complete!")
```
